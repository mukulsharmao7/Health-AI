```python
daily_data["sleep_deviation_pct"] = (
    daily_data["sleep_deviation"]
    / daily_data["baseline_sleep_minutes"]
) * 100
```

```python
daily_data = daily_data.sort_values(
    ["Id", "date"]
).reset_index(drop=True)
```

```python
daily_data["day_number"] = (
    daily_data.groupby("Id").cumcount() + 1
)
```

```python
daily_data["is_baseline"] = daily_data["day_number"] <= 7
```

```python
daily_data[["Id", "date", "day_number", "is_baseline"]].head(10)
```

```python
baseline_7d = (
    daily_data[daily_data["is_baseline"]]
    .groupby("Id")["TotalSteps"]
    .mean()
    .reset_index()
)
```

```python
daily_data[daily_data["is_baseline"]]
```

```python
baseline_7d.head()
```

```python
baseline_7d = baseline_7d.rename(
    columns={"TotalSteps": "baseline_steps_7d"}
)
```

```python
daily_data = daily_data.merge(
    baseline_7d,
    on="Id",
    how="left"
)
```

```python
daily_data["steps_deviation_7d"] = (
    daily_data["TotalSteps"] - daily_data["baseline_steps_7d"]
)
```

```python
daily_data["steps_deviation_7d_pct"] = (
    daily_data["steps_deviation_7d"]
    / daily_data["baseline_steps_7d"]
) * 100
```

```python
sleep_baseline_7d = (
    daily_data[daily_data["is_baseline"]]
    .groupby("Id")["recorded_sleep_minutes"]
    .mean()
    .reset_index()
)
```

```python
sleep_baseline_7d = sleep_baseline_7d.rename(
    columns={"recorded_sleep_minutes": "baseline_sleep_7d"}
)
```

```python
sleep_baseline_7d.head()
```

```python
daily_data = daily_data.merge(
    sleep_baseline_7d,
    on="Id",
    how="left"
)
```

```python
# ==========================================
# DAY 6 — FINAL FEATURE ENGINEERING
# ==========================================

# 1. Sleep deviation from 7-day personal baseline
daily_data["sleep_deviation_7d"] = (
    daily_data["recorded_sleep_minutes"]
    - daily_data["baseline_sleep_7d"]
)

# 2. Sleep percentage deviation
daily_data["sleep_deviation_7d_pct"] = (
    daily_data["sleep_deviation_7d"]
    / daily_data["baseline_sleep_7d"]
) * 100


# 3. Sort chronologically before rolling features
daily_data = daily_data.sort_values(
    ["Id", "date"]
).reset_index(drop=True)


# 4. 3-observation rolling average of steps
daily_data["steps_rolling_3d"] = (
    daily_data
    .groupby("Id")["TotalSteps"]
    .transform(
        lambda x: x.rolling(window=3, min_periods=1).mean()
    )
)


# 5. 3-observation rolling average of recorded sleep
daily_data["sleep_rolling_3d"] = (
    daily_data
    .groupby("Id")["recorded_sleep_minutes"]
    .transform(
        lambda x: x.rolling(window=3, min_periods=1).mean()
    )
)


# 6. Activity ratio relative to personal baseline
daily_data["activity_ratio"] = (
    daily_data["TotalSteps"]
    / daily_data["baseline_steps_7d"]
)


# 7. Sleep ratio relative to personal baseline
daily_data["sleep_ratio"] = (
    daily_data["recorded_sleep_minutes"]
    / daily_data["baseline_sleep_7d"]
)


# 8. Select final ML-ready features
feature_columns = [
    "Id",
    "date",
    "TotalSteps",
    "Calories",
    "VeryActiveMinutes",
    "FairlyActiveMinutes",
    "LightlyActiveMinutes",
    "SedentaryMinutes",
    "recorded_sleep_minutes",
    "baseline_steps_7d",
    "baseline_sleep_7d",
    "steps_deviation_7d",
    "steps_deviation_7d_pct",
    "sleep_deviation_7d",
    "sleep_deviation_7d_pct",
    "steps_rolling_3d",
    "sleep_rolling_3d",
    "activity_ratio",
    "sleep_ratio",
    "day_number",
    "is_baseline"
]

feature_data = daily_data[feature_columns].copy()


# 9. Basic validation
print("Shape:", feature_data.shape)
print("\nMissing values:")
print(feature_data.isna().sum())

print("\nDuplicate rows:")
print(feature_data.duplicated().sum())

print("\nFeature dataset preview:")
display(feature_data.head())


# 10. Save processed feature dataset
from pathlib import Path

processed_path = Path("../data/processed")
processed_path.mkdir(parents=True, exist_ok=True)

feature_data.to_csv(
    processed_path / "daily_features.csv",
    index=False
)

print("\nFeature engineering completed successfully.")
print("Saved to: ../data/processed/daily_features.csv")
```
