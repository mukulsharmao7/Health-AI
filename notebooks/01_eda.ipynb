```python
import matplotlib.pyplot as plt

participant_dates=df.groupby("Id")["ActivityDate"].agg(["min","max"])

plt.figure(figsize=(12, 8))

for i, (participant, row) in enumerate(participant_dates.iterrows()):
    plt.plot(
        [row["min"], row["max"]],
        [i, i],
        marker="o"
    )

plt.xlabel("Date")
plt.ylabel("Participant")
plt.title("Participant Data Coverage Over Time")
plt.grid(True)
plt.show()
```

```python
participant_id = 4020332650
person = df[df["Id"] == participant_id]

plt.figure(figsize=(12,5))
plt.plot(person["ActivityDate"], person["TotalSteps"], marker="o")
plt.xlabel("Date")
plt.ylabel("Total Steps")
plt.title(f"Daily Steps - Participant {participant_id}")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.hist(df["TotalSteps"], bins=30)
plt.xlabel("Total Steps")
plt.ylabel("Number of Days")
plt.title("Distribution of Daily Steps")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df["TotalSteps"], df["Calories"])
plt.xlabel("Total Steps")
plt.ylabel("Calories")
plt.title("Daily Steps vs Calories Burned")
plt.grid(True)
plt.show()
```

## EDA Summary

- The dataset contains 35 unique participants with unequal observation periods.
- Each participant has an average of approximately 13 daily observations.
- The available activity data spans approximately 12 March 2016 to 12 April 2016.
- Daily activity shows substantial variation both across participants and within the same participant over time.
- The distribution of daily steps is right-skewed, with several high-activity days.
- TotalSteps and TotalDistance show a very strong positive correlation (0.987), indicating substantial redundancy between these features.
- TotalSteps and Calories show a moderate positive correlation (0.581).
- Individual longitudinal activity patterns support the need for personalized baseline modeling rather than relying only on population-level averages.
